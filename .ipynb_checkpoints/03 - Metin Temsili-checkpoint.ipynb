````python
import re
from bs4 import BeautifulSoup as bs
from nltk.corpus import stopwords
stop_words_eng = set(stopwords.words("english"))

def clean_text(text):
    text = text.lower()

    # Rakamları temizlemek için yazdığımız regular expresion.
    text = re.sub(r"\d+", "", text)

    text = bs(text, "html.parser").get_text()
    
    # Özel karakterlerin temizlenmesi için yazdığımız regular expression. '^' = not, '\w' = harfler, '\s' = boşluk
    # Esasında "[^\w\s]" iken, SMS spam veri setinde yalnızca 'underscore' karakterinden oluşan tokenlar da oluduğu için buna çevirdim.
    text = re.sub(r"[^A-Za-z\s]", "", text)

    ttokens = text.split()
    text = " ".join([word for word in ttokens if word not in stop_words_eng and len(word)>2])

    return text
````

## 2. TF-IDF (Term Frequency-Inverse Document Frequency)

- Metin madenciliğinde ve bilgi erişiminde sıkça kullanılan bir özellik çıkarım yöntemidir.
- Kelimelerin belgeler içinde ne kadar önemli olduğunu belirlemek için kullanılır.
- **Term Frequency (TF):** Bir kelimenin bir belgede ne kadar sık geçtiğini ölçer.
- **Inverse Document Frequency (IDF):** Bir kelimenin tüm belgelerdeki yaygınlığını ölçer, kelime ne kadar çok belgede geçiyorsa o kelime çok fazla bilgi sağlamaz.

$$ \text{TF}(t,d) = \frac{\text{Sayaç}(t,d)}{\text{Toplam Kelime Sayısı}(d)}$$
<center>Burada, <b><em>Sayaç(t,d)</em></b> denilen kavram <b><em>t</em></b> kelimensinin <b><em>d</em></b> dökümanında geçme sayısıdır. <b><em>Toplam Kelime Sayısı(d)</em></b> ise <b><em>d</em></b> dökümanındaki toplem kelime sayısıdır.</center>

$$ \text{IDF}(t,D) = log(\frac{\text{Toplam Belgeler Sayısı}(D)}{1 + \text{Belgelerde Geçen Sayısı}(t)})$$
<center>Burada <b><em>Toplam Belgeler Sayısı(D)</em></b> toplam döküman sayısı, <b><em>Belgelerde Geçen Sayısı (t)</em></b> ise <b><em>t</em></b> kelimesinin geçtiği belge sayısıdır.</center>

$$ \text{TF-IDF}(t,d,D) = \text{TF}(t,d) \times \text{IDF}(t,D)$$

````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
````

## 3. N-Gram Modelleri

- Bir dil modelinde kullanılan kelime veya karakter dizisinin uzunluğunu belirten bir terimdir.
- N-Gram modelleri, metinleri <em>n</em> kelimelik veya <em>n</em> karakterlik kısımlara bölerek analiz eder.
    - <b><em>Uni-Gram:</em></b> `This` `Is` `Big` `Data` `AI` `Book`
    - <b><em>Bi-Gram:</em></b> `This is` `Is Big` `Big data` `Data AI` `AI Book`
    - <b><em>Tri-Gram:</em></b> `This is big` `Is big data` `Big data AI` `Data AI Book`
- Kullanım Alanları:
    - Metin modelleme
    - Metin Sınıflandırma
    - Metin Üretimi
    - Metin Benzerliği

````python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

documents = ["Bu çalışma NGram analiz çalışmasıdır", "Bu çalışma bir doğal dil işleme algoritması çalışmasıdır"]
vectorizer_unigram = CountVectorizer(ngram_range = (1,1))
vectorizer_bigram = CountVectorizer(ngram_range = (2,2))
vectorizer_trigram = CountVectorizer(ngram_range = (3,3))

X_unigram = vectorizer_unigram.fit_transform(documents)
unigram_features = vectorizer_unigram.get_feature_names_out()

X_bigram = vectorizer_bigram.fit_transform(documents)
bigram_features = vectorizer_bigram.get_feature_names_out()

X_trigram = vectorizer_trigram.fit_transform(documents)
trigram_features = vectorizer_trigram.get_feature_names_out()
````

## 4. Word Embeddings

- Kelimeleri, genellikle sürekli bir vektör uzayında anlamlı temsil edecek şekilde sayısal vektörlere dönüştürür.
- Bu temsiller, kelimeler arasındaki anlamsal ve dilbilgisel ilişkiler yakalamayı hedefler.
    - Anlamsal Benzerlik: Örn. `King` -> `Queen`
    - Matematiksel İşlemler (Vektör Ekleme/Çıkartma): Örn. `King` - `man` + `woman` = `Queen`
- Kapsamlılık: Büyük veri kümesiyle eğitilir, dildeki çeşitli ilişkileri yakalar.
- Word Embeddings Modelleri:
    - Word2Vec: Google tarafından geliştirilen, kelimeleri vektörlere dönüştüren ve bu vektörleri dildeki ilişkileri yakalayacak şekilde öğrenen bir modeldir.
    - GloVe (Global Vectors for Word Representation): Stanford Üniversitesi tarafından geliştirilen, kelime gömme temsillerini keli ortaklıklarını yaklayacak şekilde hesaplayan bir modeldir.
    - FastText: Facebook tarafından geliştirilen ve kelime gömme temsillerini kelime alt-birimlerini de dikkate alarak hesaplayan bir modeldir.

````python
import pandas as pd
import matplotlib.pyplot as plt

# Principle Component Analysis: Verideki önemli bilgileri korurken veri boyutunu azaltan ve yeni özellik seti oluşturan bir yöntemdir. Bunu verideki varyansı en iyi şekilde açıklayan yeni doğrusal bileşenler (principle component) oluşturarak yapar. 
from sklearn.decomposition import PCA
from gensim.models import Word2Vec, FastText
from gensim.utils import simple_preprocess
````

```python
Word2Vec(*a, vector_size = *b, window = *c, min_count = *d, sg = *e)
```

- <b>a:</b> Tokenize edilmiş cümleler.
- <b>b:</b> Kelimelerin oluşturacağı embedding vector'lerinin boyutu.
- <b>c:</b> Her bir kelimenin anlamını çıkartmak için kullanacağı diğer kelimelerin, üzerinde çalışan kelimeye olan uzaklığı.
- <b>d:</b> Eğitimde dikkate alınacak kelimelerin minimum görülme sayısı.
- <b>e:</b> Word2Vec modelin mimarisini belirler. Örneğin 0 değeri CBoW (Continuous Bag of Words), 1 ise Skip-Gram içindir.
    - CBoW: Kayıp kelimeyi bulmak için etrafındaki kelimeleri kullanır. Eğitim hızı yüksektir ve büyük verilerde sık kullanılan yaygın kelimelerde iyi çalışır.
      - `The cat ____ on the mat.` cümlesinde `[The, cat, on, the, mat]` kelimelerini kullanark `sat` kelimesini tahmin eder. 
    - Skip-Gram: Bir kelimeyi kullanarak etrafındaki kelimeleri tahmin eder. Eğitimi, tek kelimeden çok sayıda kelime tahmin edeceği için uzun sürer. En iyi, sık kullanılmayan kelimerle çalışır.
        - `sat` kelimesinden `The cat sat on the mat` cümlesini üretir.
  

````python
sentences = ["Köpek çok tatlı bir hayvandır",
             "Köpekler evcil hayvanlardır",
             "Kediler genellikle bağımsız hareket etmeyi severler",
             "Köpekler sadık ve dost canlısı hayvanlardır",
             "Evcil hayvanlar insanlar için iyi birer arkadaştırlar"]


tokenized_sentences = [simple_preprocess(sentence) for sentence in sentences]

word2_vec_model = Word2Vec(sentences = tokenized_sentences, vector_size = 50, window = 5, min_count = 1, sg = 0)
fasttext_model = FastText(sentences = tokenized_sentences, vector_size = 50, window = 5, min_count = 1, sg = 0)
````

````python
def plot_word_embedding(model, title):
    word_vectors = model.wv

    words = list(word_vectors.index_to_key)[:1000]
    vectors = [word_vectors[word] for word in words]

    # PCA
    pca = PCA(n_components=3)
    reduced_vectors = pca.fit_transform(vectors)

    # 3D Görselleştirme
    fig = plt.figure(figsize = (8,6))
    ax = fig.add_subplot(111, projection = "3d")

    # Vektörlerin çizilmesi
    ax.scatter(reduced_vectors[:,0], reduced_vectors[:,1], reduced_vectors[:,2])

    # Kelimelerin etiketlenmesi
    for i, word in enumerate(words):
        ax.text(reduced_vectors[i,0], reduced_vectors[i,1], reduced_vectors[i,2], word, fontsize=12)

    ax.set_title(title)
    ax.set_xlabel("Component 1")
    ax.set_ylabel("Component 2")
    ax.set_zlabel("Component 3")
````

````python
plot_word_embedding(word2_vec_model, "Word2Vec")
plot_word_embedding(fasttext_model, "FastText")
````
