### Boşlukların Temizlenmesi:

````python
text = "Hello,          World,              2026"
cleanedtext = " ".join(text.split())
print(f"Text: {text}\nCleaned Text: {cleanedtext}")
````

### Büyük-Küçük Harf Dönüşümleri:

````python
print(f"All lower: {text.lower()}\nAll upper: {text.upper()}")
````

### Noktalama İşaretlerinin Kaldırılması:

````python
import string


# İlk iki parametreyi boş girmemizin sebebi herhangi bir karakteri başka bir karakterle değiştirmek istemiyor olmamız
text.translate(str.maketrans("", "", string.punctuation))
````

### Özel Karakterlerin Kaldırılması:

````python
import re
text2 = "Hello,   %%  @@     World,    // **       2026## - ğ ü ş İ ö ç"
cleanedText = re.sub(r"[^A-Za-z0-9\s]", "", text2)
textNoPunctuation = text2.translate(str.maketrans("", "", string.punctuation))
print(f"Original: {text2}\nCleaned With Punctuation: {textNoPunctuation}\nCleaned Text: {cleanedText}")
````
