## Tokenizasyon

- Tokenizasyon, bir metni daha küçük parçalara ayırma işlemidir.
- Tokenlar; kelimeler, cümleler ve hatta karakterler olabilir.

````python
import nltk

nltk.download("punkt") # Metni kelime ve cümle bazında tokenlara ayırmak için gerekli.
nltk.download("punkt_tab") # Metni kelime ve cümle bazında tokenlara ayırmak için gerekli.
````

````python
text = "Hello, World! How are you? Hello, hi..."

# Kelime tokenizasyonu: word_tokenize -> Noktalama işaretleri ve boşuklar da ayrı birer token olarak elde edilir.
word_tokens = nltk.word_tokenize(text)

# Cümle tokenizasyonu: sent_tokenize -> Her cümle bir token
sent_tokens = nltk.sent_tokenize(text)

print(word_tokens)
print(sent_tokens)
````

## Kök ve Gövde Analizi (Stemming and Lemmatization)

- **Stemming (Kök Bulma)**
    - Kelimelerin kök formunu bulmak için kelime sonundaki eklerin (suffix) çıkartılması.
    - Kelimenin anlamını tamamen doğru bir şekilde elde etmeyi amaçlamaz, kelimenin en basit formuna odaklanır.
        - *Koşuyor*, *Koştu*, *Koşmak* -> *Koş*
        - *Evde*, *Evler*, *Evimiz* -> *Ev*
<br></br>
- **Göve Analizi (Lemmatization)**
    - Kelimeleri sözlükteki temel formlara (lemma) dönüştürme işlemidir.
    - Kelimenin anlamını ve dilbilgisel yapısını dikkate alarak doğru bir kök bulmaya çalışır.
    - Bu nedenle, lemmazitation sonrası elde edilen kelime dilbilgisel oalrak anlamlı ve zölükte yer alan bir kelime olur.
        - *Koşuyor*, *Koştu*, *Koşmak* -> *Koşmak*
        - *Evde*, *Evler*, *Evimiz* -> *Ev*

````python
nltk.download("wordnet") # Lemmatization işlemi için gerekli

from nltk.stem import PorterStemmer
````

````python
words = ["running", "runner", "runs", "better", "go", "went"]
````

````python
stemmer = PorterStemmer()

[stemmer.stem(w) for w in words]
````

````python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
````

````python
[lemmatizer.lemmatize(w) for w in words]
````

Yukarıda istenilen sonuçları elde edememizin nedeni fiil, sıfat ve isimleri bir arada vermiş olmamız. Eğer `lemmatize()` fonksiyonuna, verdiğimiz listeyi fiil olarak işaretleseydik daha iyi sonuçlar alınabilirdi.

````python
[lemmatizer.lemmatize(w, pos="v") for w in words]
````

## Durdurma Kelimeleri (Stop Words)

- Durdurma kelimeleri, metinlerde genellikle anlamı çok az olan veya metnin analizi sırasında çok faydalı olmayan kelimelerdir **(bağlaçlar, edatlar, zamirler vb.)**.
- Bu kelimelerin kaldırılması analiz doğruluğunu artırabilir.
    - *And, The, is, in, to, of, it, that*

````python
from nltk.corpus import stopwords
nltk.download("stopwords")
````

````python
# Stop Words (nltk)
stop_words_eng = set(stopwords.words("english"))

text = "There are multiple examples of handling and removing stop words from some text."
text_list = text.lower().split()

filtered_list = [word for word in text_list if word not in stop_words_eng]
filtered_list
````

````python
# Durdurma Kelimeleri (nltk)
durdurma_kelimeleri = set(stopwords.words("turkish"))
metin = "Merhabalar sizler ile çok güzel bir ders işliyoruz."

metin_liste = metin.lower().split()
filtreli_liste = [kelime for kelime in metin_liste if kelime not in durdurma_kelimeleri]
filtreli_liste
````
